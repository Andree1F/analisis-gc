```python
# Celda 1 — importar herramientas
from Bio.Seq import Seq
from Bio import SeqIO
import matplotlib.pyplot as plt

print("✓ Librerías cargadas correctamente")
```

```python
# Celda 2 — crear nuestras secuencias de prueba
secuencias = {
    "gen_BRCA1": Seq("ATGGATTTTTGTTTTAGTTTTTCCTGATGCAGATTTTCAGTTCAGTGGCACAGCAGATG"),
    "gen_TP53":  Seq("ATGGAGGAGCCGCAGTCAGATCCTAGCGTTGAATCCAGGGGATCCGTCCGGGTTTTGG"),
    "gen_MYC":   Seq("ATGCCCCTCAACGTTAGCTTCACCAACAGGAACTATGACCTCGACTACGACTCGGTGCA")
}

for nombre, seq in secuencias.items():
    gc = round((seq.count("G") + seq.count("C")) / len(seq) * 100, 1)
    print(f"{nombre}: {len(seq)} pb | GC: {gc}%")
```

```python
# Celda 3 — visualizar contenido GC
nombres = list(secuencias.keys())
gc_values = [round((s.count("G") + s.count("C")) / len(s) * 100, 1) 
             for s in secuencias.values()]

plt.figure(figsize=(8, 4))
colores = ["#E8593C" if gc > 50 else "#3B8BD4" for gc in gc_values]
barras = plt.bar(nombres, gc_values, color=colores, edgecolor="none")
plt.axhline(y=50, color="gray", linestyle="--", linewidth=1, alpha=0.7)
plt.ylabel("Contenido GC (%)")
plt.title("Comparación de contenido GC — genes humanos")
plt.ylim(0, 100)

for barra, val in zip(barras, gc_values):
    plt.text(barra.get_x() + barra.get_width()/2, val + 1.5,
             f"{val}%", ha="center", fontsize=11)

plt.tight_layout()
plt.savefig("grafico_gc.png", dpi=150)
plt.show()
print("✓ Gráfico guardado como grafico_gc.png")
```
